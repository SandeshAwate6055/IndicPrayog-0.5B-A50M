#### Step 1: Format Standardization (example)

Sangraha, Wikipedia and FineWeb-Edu each have a different schema. For each source below we first look at the old (raw) format, then convert it into the new canonical shape: `id, text, language, source, domain, url, title, metadata`.

```python
import pandas as pd
from pathlib import Path

DATASET_ROOT = Path("../../../../dataset").resolve()
```

#### Sangraha: old format (raw)

```python
raw = pd.read_parquet(DATASET_ROOT / "sangraha/verified/hin/data-0.parquet").iloc[0]
print("raw columns:", list(raw.index))
raw
```

#### Sangraha: new format (canonical)

```python
canonical = {
    "id": f"sangraha_hi_{raw['doc_id']}",
    "text": raw["text"],
    "language": "hi",
    "source": "sangraha",
    "domain": raw["type"],
    "url": None,
    "title": None,
    "metadata": "{}",
}
canonical
```

#### Wikipedia: old format (raw)

```python
raw = pd.read_parquet(DATASET_ROOT / "wikipedia/20231101.hi/train-00000-of-00002.parquet").iloc[0]
print("raw columns:", list(raw.index))
raw
```

#### Wikipedia: new format (canonical)

```python
canonical = {
    "id": f"wikipedia_hi_{raw['id']}",
    "text": raw["text"],
    "language": "hi",
    "source": "wikipedia",
    "domain": "encyclopedia",
    "url": raw["url"],
    "title": raw["title"],
    "metadata": "{}",
}
canonical
```

#### FineWeb-Edu: old format (raw)

```python
raw = pd.read_parquet(DATASET_ROOT / "fineweb-edu/sample/10BT/000_00000.parquet").iloc[0]
print("raw columns:", list(raw.index))
raw
```

#### FineWeb-Edu: new format (canonical)

```python
import json

canonical = {
    "id": f"fineweb-edu_en_{raw['id']}",
    "text": raw["text"],
    "language": "en",
    "source": "fineweb-edu",
    "domain": "web",
    "url": raw["url"],
    "title": None,
    "metadata": json.dumps({"language_score": raw["language_score"], "token_count": int(raw["token_count"]), "score": raw["score"]}),
}
canonical
```
