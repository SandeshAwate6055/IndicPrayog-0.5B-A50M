# Hindi Dataset Overview

Sources: `ai4bharat/sangraha` (verified, hin split) and `wikimedia/wikipedia` (20231101.hi).

Data location: `dataset/` (sibling folder to the repo root)

```python
import pandas as pd
from pathlib import Path

DATASET_ROOT = Path("../../../dataset").resolve()
pd.set_option("display.max_colwidth", 120)
```

## Sangraha (verified, Hindi)

```python
hi_sangraha = pd.read_parquet(DATASET_ROOT / "sangraha/verified/hin/data-0.parquet")
print("Sangraha Hindi (1 of many shards):", hi_sangraha.shape)
hi_sangraha.head()
```

```python
print("Sample Hindi text (Sangraha):\n")
print(hi_sangraha.iloc[0]["text"][:500])
```

```python
hi_sangraha["type"].value_counts()
```

## Wikipedia (Hindi)

```python
hi_wiki = pd.read_parquet(DATASET_ROOT / "wikipedia/20231101.hi/train-00000-of-00002.parquet")
print("Wikipedia Hindi (1 of 2 shards):", hi_wiki.shape)
hi_wiki[["title", "text"]].head()
```

```python
print("Sample Hindi text (Wikipedia):\n")
print(hi_wiki.iloc[0]["text"][:500])
```

## Summary

```python
summary = pd.DataFrame([
    {"source": "Sangraha verified", "rows": len(hi_sangraha), "avg_chars": hi_sangraha["text"].str.len().mean()},
    {"source": "Wikipedia", "rows": len(hi_wiki), "avg_chars": hi_wiki["text"].str.len().mean()},
])
summary
```
