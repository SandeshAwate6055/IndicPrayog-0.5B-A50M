# English Dataset Overview

Sources: `HuggingFaceFW/fineweb-edu` (sample/10BT) and `wikimedia/wikipedia` (20231101.en).

Data location: `dataset/` (sibling folder to the repo root)

```python
import pandas as pd
from pathlib import Path

DATASET_ROOT = Path("../../../dataset").resolve()
pd.set_option("display.max_colwidth", 120)
```

## FineWeb-Edu (English)

```python
en_fineweb = pd.read_parquet(DATASET_ROOT / "fineweb-edu/sample/10BT/000_00000.parquet")
print("FineWeb-Edu English (1 of 7 downloaded shards):", en_fineweb.shape)
en_fineweb[["text", "url", "language_score", "token_count", "score"]].head()
```

```python
print("Sample English text (FineWeb-Edu):\n")
print(en_fineweb.iloc[0]["text"][:500])
```

```python
en_fineweb["score"].describe()
```

## Wikipedia (English)

```python
en_wiki = pd.read_parquet(DATASET_ROOT / "wikipedia/20231101.en/train-00000-of-00041.parquet")
print("Wikipedia English (1 of 5 downloaded shards):", en_wiki.shape)
en_wiki[["title", "text"]].head()
```

```python
print("Sample English text (Wikipedia):\n")
print(en_wiki.iloc[0]["text"][:500])
```

## Summary

```python
summary = pd.DataFrame([
    {"source": "FineWeb-Edu", "rows": len(en_fineweb), "avg_chars": en_fineweb["text"].str.len().mean()},
    {"source": "Wikipedia", "rows": len(en_wiki), "avg_chars": en_wiki["text"].str.len().mean()},
])
summary
```
