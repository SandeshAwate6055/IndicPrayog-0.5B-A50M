# Marathi Dataset Overview

Sources: `ai4bharat/sangraha` (verified, mar split) and `wikimedia/wikipedia` (20231101.mr).

Data location: `dataset/` (sibling folder to the repo root)

```python
import pandas as pd
from pathlib import Path

DATASET_ROOT = Path("../../../dataset").resolve()
pd.set_option("display.max_colwidth", 120)
```

## Sangraha (verified, Marathi)

```python
mr_sangraha = pd.read_parquet(DATASET_ROOT / "sangraha/verified/mar/data-0.parquet")
print("Sangraha Marathi (1 of many shards):", mr_sangraha.shape)
mr_sangraha.head()
```

```python
print("Sample Marathi text (Sangraha):\n")
print(mr_sangraha.iloc[0]["text"][:500])
```

```python
mr_sangraha["type"].value_counts()
```

## Wikipedia (Marathi)

```python
mr_wiki = pd.read_parquet(DATASET_ROOT / "wikipedia/20231101.mr/train-00000-of-00001.parquet")
print("Wikipedia Marathi (full):", mr_wiki.shape)
mr_wiki[["title", "text"]].head()
```

```python
print("Sample Marathi text (Wikipedia):\n")
print(mr_wiki.iloc[0]["text"][:500])
```

## Summary

```python
summary = pd.DataFrame([
    {"source": "Sangraha verified", "rows": len(mr_sangraha), "avg_chars": mr_sangraha["text"].str.len().mean()},
    {"source": "Wikipedia", "rows": len(mr_wiki), "avg_chars": mr_wiki["text"].str.len().mean()},
])
summary
```
